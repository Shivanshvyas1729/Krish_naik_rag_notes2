### State Schema With DataClasses & TypedDict

When we define a LangGraph `StateGraph`, we specify a **state schema**.

The state schema represents the structure and types of data that our graph will use. All nodes are expected to read from and return updates conforming to that schema.

LangGraph offers flexibility in how you define your state schema, accommodating:
1. `TypedDict` (dictionary-based, static type checking)
2. `@dataclass` (object-based, attribute access with default values)
3. `Pydantic` `BaseModel` (runtime data validation & coercion)


### Part 1: State Schema using `TypedDict`

`TypedDict` from Python's standard `typing` module allows you to specify keys and expected types for dictionary-based state.

> [!NOTE]
> `TypedDict` type annotations are type hints. They help IDEs and static analyzers (like `mypy`), but are not enforced at runtime.


```python
from typing import TypedDict, Literal, Optional

class TypedDictState(TypedDict):
    name: str
    game: Optional[Literal["cricket", "badminton"]]
```

```python
def play_game(state: TypedDictState) -> dict:
    print("--- play_game node has been called ---")
    return {"name": state['name'] + " wants to play"}

def cricket(state: TypedDictState) -> dict:
    print("--- cricket node has been called ---")
    return {"name": state["name"] + " cricket", "game": "cricket"}

def badminton(state: TypedDictState) -> dict:
    print("--- badminton node has been called ---")
    return {"name": state["name"] + " badminton", "game": "badminton"}
```

```python
import random

def decide_play(state: TypedDictState) -> Literal["cricket", "badminton"]:
    # 50/50 split between cricket and badminton
    if random.random() < 0.5:
        return "cricket"
    else:
        return "badminton" 
```

```python
from IPython.display import Image, display
from langgraph.graph import StateGraph, START, END

builder = StateGraph(TypedDictState)
builder.add_node("playgame", play_game)
builder.add_node("cricket", cricket)
builder.add_node("badminton", badminton)

## Flow of the graph
builder.add_edge(START, "playgame")
builder.add_conditional_edges("playgame", decide_play)
builder.add_edge("cricket", END)
builder.add_edge("badminton", END)

# Compile
graph = builder.compile()

# View
try:
    display(Image(graph.get_graph().draw_mermaid_png()))
except Exception:
    print(graph.get_graph().draw_mermaid())
```

```python
result = graph.invoke({"name": "Krish"})
print("Result with valid string input:\n", result)
```

### Part 2: State Schema using Python `dataclasses`

Python's `dataclasses` module provides another clean way to define structured state.

**Key features with Dataclasses:**
- Provides object-style attribute access: `state.name` instead of `state["name"]`.
- Allows defining default field values (e.g. `game: Optional[...] = None`).
- Nodes can read fields using dot-notation (`state.name`) and return dictionary updates (e.g. `{"name": ...}`).


```python
from dataclasses import dataclass
from typing import Literal, Optional

@dataclass
class DataClassState:
    name: str
    game: Optional[Literal["badminton", "cricket"]] = None
```

```python
def play_game_dc(state: DataClassState) -> dict:
    print("--- play_game node (Dataclass) has been called ---")
    # State is accessed via dot-notation attributes
    return {"name": state.name + " wants to play"}

def cricket_dc(state: DataClassState) -> dict:
    print("--- cricket node (Dataclass) has been called ---")
    return {"name": state.name + " cricket", "game": "cricket"}

def badminton_dc(state: DataClassState) -> dict:
    print("--- badminton node (Dataclass) has been called ---")
    return {"name": state.name + " badminton", "game": "badminton"}
```

```python
def decide_play_dc(state: DataClassState) -> Literal["cricket", "badminton"]:
    if random.random() < 0.5:
        return "cricket"
    else:
        return "badminton" 
```

```python
builder_dc = StateGraph(DataClassState)
builder_dc.add_node("playgame", play_game_dc)
builder_dc.add_node("cricket", cricket_dc)
builder_dc.add_node("badminton", badminton_dc)

builder_dc.add_edge(START, "playgame")
builder_dc.add_conditional_edges("playgame", decide_play_dc)
builder_dc.add_edge("cricket", END)
builder_dc.add_edge("badminton", END)

graph_dc = builder_dc.compile()

try:
    display(Image(graph_dc.get_graph().draw_mermaid_png()))
except Exception:
    print(graph_dc.get_graph().draw_mermaid())
```

```python
# Invoking by passing a DataClass instance
out_instance = graph_dc.invoke(DataClassState(name="Krish"))
print("Result when passing DataClass instance:\n", out_instance)
```

```python
# Invoking by passing a standard dictionary (LangGraph initializes the dataclass automatically)
out_dict = graph_dc.invoke({"name": "Krish"})
print("Result when passing dictionary:\n", out_dict)
```
